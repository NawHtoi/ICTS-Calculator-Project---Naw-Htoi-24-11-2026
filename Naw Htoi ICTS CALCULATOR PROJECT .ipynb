## Project Description

This project is to build a basic calculator that can perform addition, subtraction, multiplication, division, second power, and square root.
The user can pick the operations from a menu and repeat this action until they wish to stop.
Each operation is implemented as its own function for convenience and simplicity.

### First Part

In the code cell below, I will be creating basic operations such as addition, subtraction, multiplication, and division. The only noteworthy thing I did here was adding in an error message in case someone tries to divide by zero.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
```

In the next code cell, I will be implementing the harder operations, square root and second power. Similarly to the divide by zero error, I had to put in an error message for someone trying to get a square root of a negative number. Because both these errors are already built into Jupyter/Python, it was very easy to do.

```python
def second_power(a):
    return a ** 2

def square_root(n):
    if n < 0:
        raise ValueError("Cannot take square root of a negative number")
    return n ** 0.5
```

### Second Part

Within the following cells, I will create the menu, which will be a very simple menu that asks the user what operation they would like to use, before asking the two numbers, or in the case of Square Root just one number, and then showing the result. I will also add in the exit function in the next code cell.

```python
def menu():
    " Menu "
    print(" Calculator ")
    print("1. Addition ")
    print("2. Subtraction ")
    print("3. Multiplication ")
    print("4. Division ")
    print("5. Second power ")
    print("6. Square root ")
    print("7. Exit ")
```

```python
def calc():
    " Calculator "
    while True:
        menu()
        choice = input("Choose an option: ")

        if choice == '7':
            print("Exited")
            break

        try:
            if choice == '6':
                n = float(input("Enter a number: "))
                print(f"Square root of {n} = {square_root(n)}")

            elif choice == '5':
                a = float(input("Enter a number: "))
                print(f"Second power of {a} = {second_power(a)}")

            elif choice in ('1', '2', '3', '4'):
                a = float(input("First number:  "))
                b = float(input("Second number: "))
                operations = { '1': add, '2': subtract, '3': multiply, '4': divide }
                print(f"Result: {operations[choice](a, b)}")

            else:
                print("Please pick between options 1 to 7.")

        except (ValueError, ZeroDivisionError) as e:
            print(f"Error: {e}")
```
